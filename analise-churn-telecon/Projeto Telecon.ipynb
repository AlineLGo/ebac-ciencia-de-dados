```python
import pandas as pd
```

```python
# 1. Carregando a base de dados indicando o separador correto (ponto e vírgula)
df = pd.read_csv('/CHURN_TELECON_MOD08_TAREFA.csv', sep=';')
# 2. Mostrando as 5 primeiras linhas da tabela para ver se carregou certinho
display(df.head())
# 3. Verificando o tipo de dado de cada coluna
df.info()
```

```python
# 1. Alterando o tipo de dado da coluna 'Idoso' de número inteiro para texto
df['Idoso'] = df['Idoso'].astype(str)
# 2. Verificando novamente para confirmar se a mudança aconteceu
df.info()
```

```python
# 1. Contando quantos dados estão faltando (nulos) em cada coluna
valores_ausentes = df.isnull().sum()
# 2. Calculando o percentual de dados faltantes (quantidade de nulos dividida pelo total de linhas, multiplicando por 100)
percentual_ausentes = (df.isnull().sum()/len(df)) * 100
# 3. Juntando as duas informações em uma tabelinha bonita para visualização
tabela_nulos = pd.DataFrame({'Valores Ausentes': valores_ausentes, 'Percentual (%)': percentual_ausentes})
# 4. Mostrando a tabelateF na tela, ornadenada do maior problema para o menor
display(tabela_nulos.sort_values(by='Percentual (%)', ascending=False))
```

```python
# 1. Excuindo a coluna 'PhoneService' inteira por ter quase 60% de dados ausentes
df = df.drop(columns=['PhoneService'])
# 2. Excluindo apenas as linhas onde a coluna 'Churn' está vazia
df = df.dropna(subset=['Churn'])
# 3. Verificando o tamanho da tabela após as exclusões
print(f"Tamanho da base atual:{df.shape}")
```

```python
# 1. Preenchendo os valores do 'Pagamento_Mensal' com MEDIANA
mediana_pagamento = df['Pagamento_Mensal'].median()
df['Pagamento_Mensal'].fillna(mediana_pagamento, inplace=True)
# 2. Preenchendo os valores do 'Genero' com a MODA
moda_genero = df['Genero'].mode()[0]
df['Genero'].fillna(moda_genero, inplace=True)
# 3. Fazendo o 'Raio-X' final para garantir que não há mais nenhum dado vazio na tabela
print(df.isnull().sum())
```

```python
# 1. Padronizando a coluna 'Genero'
# Tudo o que for 'F' ou 'f' vira 'Female', e o que for 'M' vira 'Male'
df['Genero'] = df['Genero'].replace({'F':'Female','f':'Female', 'M': 'Male'})
# 2. Padronizando a coluna 'Servico_Internet'
# Tudo o que for 'dsl' minúsculo vira 'DSL' maiúsculo
df['Servico_Internet'] = df['Servico_Internet'].replace({'dsl':'DSL'})
# 3. Verificando se os valores únicos ficaram limpos
print("Valores únicos em Gênero:", df['Genero'].unique())
print("Valores únicos em Internet:", df['Servico_Internet'].unique())
df
```

```python
# 1. Renomeando as colunas para o Português e padrão snake_case
df = df.rename(columns={
    'customerID': 'id_cliente',
    'Genero': 'genero',
    'Idoso': 'idoso',
    'Casado': 'casado',
    'Dependents': 'dependentes',
    'Tempo_como_Cliente': 'tempo_como_cliente',
    'Servico_Internet': 'servico_internet',
    'Servico_Seguranca': 'servico_seguranca',
    'Suporte_Tecnico': 'suporte_tecnico',
    'StreamingTV': 'streaming_tv',
    'Tipo_Contrato': 'tipo_contrato',
    'PaymentMethod': 'metodo_pagamento',
    'Pagamento_Mensal': 'pagamento_mensal',
    'Total_Pago': 'total_pago',
    'Churn': 'churn'
})

# 2. Exibindo as 5 primeiras linhas para vermos a tabela final, pronta e bonita!
display(df.head())
```
