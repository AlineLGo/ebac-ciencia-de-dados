Abaixo temos um dataframe com as seguintes informações:


**Nome:** O nome de cada indivíduo da base


**Idade:** A idade de cada um deles


**Profissão:** A atual ocupação


**Salário:** O salário registrado


**Limite de crédito:** O limite de crédito disponível para cada um dos indivíduos


**Historico_Inadimplencia:** Marco como 0 aqueles que nunca tiveram um histórico de inadimplência e como 1 aqueles que já tiveram em algum momento.


**Estado civil:** O estado civil de cada um deles


**Imovel_Proprio:** Marcado como 0 aqueles que não tem imóvel próprio e como 1 aqueles que tem.

```python
import pandas as pd
from prompt_toolkit.output import plain_text
```

```python

data = {
    'Nome': ['Alice', 'Joao', 'Charlie', 'David', 'Eva', 'Diego', 'Denize', 'Claudio'],
    'Idade': [25, 30, 35, 40, 45, 60, 22, 24],
    'Profissão': ['Engenheiro', 'Médico', 'Professor', 'Advogado', 'Médico','Engenheiro', 'Estudante','Estudante'],
    'Salário': ['4500', '8000', '5000', '10000', '12000','15000', '1200','1500'],
    'Limite_Credito': ['2500', '4000', '4000', '1000', '10000','2000', '500','250'],
    'Historico_Inadimplencia': ['0', '0', '0', '1', '0','1', '0','1'],
    'Estado_Civil': ['Casamento', 'Casamento', 'Solteiro', 'Solteiro', 'Casamento','Solteiro', 'Solteiro','Solteiro'],
    'Imovel_Proprio': ['0', '0', '0', '1', '1','1', '0','0']
}

df = pd.DataFrame(data)
df.head(8)
```

```python
# Transformando as colunas 'Salário', 'Limite_Credito' e Historico_Inadimplencia' em colunas de números inteiros
df[['Salário', 'Limite_Credito', 'Historico_Inadimplencia']]=df[['Salário', 'Limite_Credito', 'Historico_Inadimplencia']].astype(int)
df.head(8)
```

```python
# Pegando os 3 maiores limites de crédito e selecionando as colunas específicas
top_3_limites = df.nlargest(3, 'Salário')[['Nome','Salário', 'Limite_Credito','Historico_Inadimplencia' ]]
print(top_3_limites)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
# Definindo o estilo visual e o tamanho da imagem
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10,6))
# Criando o gráfico de dispersão
sns.scatterplot(
    data=df,
    x='Salário',
    y='Limite_Credito',
    hue='Historico_Inadimplencia',
    palette='deep', # Define um esquema de cores elegante
    s=150 # Aumenta o tamanho das bolinhas para facilitar a leitura
)
# Adicionando título
plt.title('Gráfico 1: Impacto do Salário e de Inadimplência no Limite de Crédito', fontsize=14)
plt.show()
```

```python
# Definindo o estilo visual e o tamanho da imagem
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10,6))
# Criando o gráfico de dispersão
sns.scatterplot(
    data=df,
    x='Idade',
    y='Limite_Credito',
    hue='Historico_Inadimplencia',
    palette='deep', # Define um esquema de cores elegante
    s=150 # Aumenta o tamanho das bolinhas para facilitar a leitura
)
# Adicionando título
plt.title('Gráfico 1: Impacto da Idade e de Inadimplência no Limite de Crédito', fontsize=14)
plt.show()
```

```python
# Gráfico 2: Qual profissão tem os maiores limites?
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10,6))
sns.barplot(
    data=df,
    x='Profissão',
    y='Limite_Credito',
    hue='Historico_Inadimplencia', # Mantemos a cor para ver quem derruba a média!
    palette='muted',
    errorbar=None
)
plt.title('Gráfico 2: Limite de Crédito por Profissão e Histórico de Pagamento', fontsize=14)
plt.show()
```

```python
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10,6))
sns.barplot(
    data=df,
    x='Imovel_Proprio',
    y='Limite_Credito',
    hue='Historico_Inadimplencia', # Mantemos a cor para ver quem derruba a média!
    palette='muted',
    errorbar=None
)
plt.title('Gráfico 2: Limite de Crédito por Imovel e Histórico de Pagamento', fontsize=14)
plt.show()
```

```python
import plotly.express as px
# Criando um gráfico interativo de pontos (strip)
fig = px.strip(
    df,
    x='Imovel_Proprio',
    y='Limite_Credito',
    color='Historico_Inadimplencia',
    hover_data=['Nome','Salário'], # Informações aparecem quando passa o mouse
    title='Gráfico 3 Interativo: Imóvel vs Limite',
    stripmode='group'
)
# Aumentando o tamanho das bolinhas
fig.update_traces(marker=dict(size=12))
fig.show()
```
